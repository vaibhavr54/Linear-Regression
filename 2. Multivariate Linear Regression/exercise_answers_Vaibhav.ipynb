```python
import numpy as np
import pandas as pd
from sklearn import linear_model
```

```python
df = pd.read_csv('hiring.csv')
```

```python
df
```

```python
df['experience'].fillna('zero', inplace=True)
df
```

```python
df.info()
```

```python
df['test_score(out of 10)'].fillna(df['test_score(out of 10)'].mean(), inplace = True)
df.info()
```

```python
df
```

```python
df['test_score(out of 10)'] = df['test_score(out of 10)'].round(0).astype('int')
```

```python
df
```

```python
from word2number import w2n
df.experience = df.experience.apply(w2n.word_to_num)
df.head()
```

```python
lr = linear_model.LinearRegression()
```

```python
lr.fit(df.drop('salary($)', axis='columns'), df[['salary($)']])
```

```python
lr.coef_
```

```python
lr.intercept_
```

```python
lr.predict([[2,9,6]])
```
