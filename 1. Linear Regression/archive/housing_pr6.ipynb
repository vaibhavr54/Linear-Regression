```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('housing.csv')
```

```python
df.sample(10)
```

```python
df1 = df.drop(['longitude', 'latitude', 'households', 'median_income', 'ocean_proximity'], axis='columns')
```

```python
df1
```

```python
df1.rename(columns={'housing_median_age': 'age', 'median_house_value': 'price'}, inplace=True)
```

```python
df1
```

```python
df1.describe()
```

```python
df1.isnull().sum()
```

```python
df1['total_bedrooms'].fillna((df1['total_bedrooms']).mean(), inplace=True)
```

```python
X = df1.drop('price', axis='columns')
```

```python
y = df1.price
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model.fit(X_train, y_train)
```

```python
model.score(X_test, y_test)
```
