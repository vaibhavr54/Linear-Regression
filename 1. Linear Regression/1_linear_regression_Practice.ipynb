```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('homeprices.csv')
df.head()
```

```python
plt.scatter(df.area, df.price, marker='+', color='red')
plt.xlabel('Area in sqft')
plt.ylabel('Price')
```

```python
from sklearn import linear_model
lr = linear_model.LinearRegression()
lr.fit(df[['area']], df[['price']])
```

```python
plt.scatter(df.area, df.price, marker='+', color='red')
plt.plot(df.area, lr.predict(df[['area']]))
plt.xlabel('Area in sqft')
plt.ylabel('Price')
```

```python
lr.predict([[3300]])
```

```python
lr.predict([[5000]])
```

```python
lr.coef_
```

```python
lr.intercept_
```

```python
dataframe = pd.read_csv('areas.csv')
dataframe.head()
```

```python
dataframe['price'] = lr.predict(dataframe)
```

```python
dataframe.to_csv('Predicted_Prices', index = False)
```
